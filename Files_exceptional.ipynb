```python
 #4 Write a Python script that reads from one file and writes its content to another file

def copy_file_content(source_file, destination_file):
    """Reads content from a source file and writes it to a destination file.

    Args:
        source_file: The path to the source file.
        destination_file: The path to the destination file.
    """
    try:
        with open(source_file, 'r') as infile, open(destination_file, 'w') as outfile:
            for line in infile:
                outfile.write(line)
    except FileNotFoundError:
        print(f"Error: Source file '{source_file}' not found.")
    except Exception as e:
        print(f"An error occurred: {e}")

# Example usage (replace with your file paths):
copy_file_content('input.txt', 'output.txt')
```

```python
#8 Write a program to handle a file opening error using exception handling

import logging

def handle_file_errors(filename):
    try:
        with open(filename, 'r') as file:
            contents = file.read()
            print(contents)
    except FileNotFoundError:
        logging.error(f"Error: File '{filename}' not found.")
    except Exception as e:
        logging.exception(f"An unexpected error occurred: {e}")

# Configure logging (optional)
logging.basicConfig(filename='file_errors.log', level=logging.ERROR,
                    format='%(asctime)s - %(levelname)s - %(message)s')


handle_file_errors("nonexistent_file.txt")
handle_file_errors("example.txt") #example.txt needs to exist
```

```python
#9 How can you read a file line by line and store its content in a list in Python?

def read_file_into_list(filepath):

    try:
        with open(filepath, 'r') as file:
            lines = file.readlines()
            return lines
    except FileNotFoundError:
        print(f"Error: File '{filepath}' not found.")
        return []
    except Exception as e:
        print(f"An error occurred while reading the file: {e}")
        return []
```
